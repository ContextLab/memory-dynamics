```python
import quail
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import hypertools as hyp

%matplotlib inline
```

```python
sns.set_context('talk')
```

```python
egg = quail.load('example')
```

```python
spc = egg.analyze('spc', listgroup=['average']*8)
```

```python
spc.plot(title='Serial Position Curve', ylim=(.2,1), legend=False)
plt.xticks(np.arange(0, 15, 2))
plt.xlabel('Presentation position')
plt.ylabel('Proportion recalled')
plt.tight_layout()
plt.savefig('figures/spc.pdf')
plt.show()
```

```python
pfr = egg.analyze('pfr', listgroup=['average']*8)
```

```python
pfr.plot(title='Probability of First Recall', ylim=(-.01,.5), legend=False)
plt.xticks(np.arange(0, 15, 2))
plt.xlabel('Presentation position')
plt.ylabel('Recall probability:\nposition 0')
plt.tight_layout()
plt.savefig('figures/pfr.pdf')
plt.show()
```

```python
lagcrp = egg.analyze('lagcrp', listgroup=['average']*8)
```

```python
lagcrp.plot(title='Lag-CRP', legend=False)
# plt.xticks(np.arange(0, 15, 2))
plt.ylim(0,.3)
plt.xlabel('Lag')
plt.ylabel('Conditional Response\nProbability')
plt.tight_layout()
plt.savefig('figures/lagcrp.pdf')
plt.show()
```
