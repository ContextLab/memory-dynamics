### 1. Setup + Basic Tokenizer 

```python
import re
import string
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

# Simple Python tokenizer
def preprocess(text):
    text = text.lower()
    text = re.sub(r"[\n\r]", " ", text)
    text = re.sub(f"[{re.escape(string.punctuation)}]", "", text)
    tokens = text.split()
    basic_stopwords = {
        'the', 'and', 'to', 'of', 'a', 'in', 'i', 'that', 'it', 'on', 'is', 'with',
        'as', 'for', 'was', 'this', 'but', 'are', 'so', 'at', 'if', 'they', 'you'
    }
    return [t for t in tokens if t not in basic_stopwords and len(t) > 2]
```

### 3. Preprocess the Text

```python
tokens_immediate = preprocess(immediate)
tokens_delayed = preprocess(delayed)
tokens_prediction = preprocess(prediction)
```

### 4. Cosine Similarity

```python
tfidf = TfidfVectorizer().fit([immediate, delayed, prediction])
tfidf_matrix = tfidf.transform([immediate, delayed, prediction])
cos_sim_id = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])[0][0]
cos_sim_ip = cosine_similarity(tfidf_matrix[0], tfidf_matrix[2])[0][0]
cos_sim_dp = cosine_similarity(tfidf_matrix[1], tfidf_matrix[2])[0][0]

print(f"Immediate vs Delayed: {cos_sim_id:.3f}")
print(f"Immediate vs Prediction: {cos_sim_ip:.3f}")
print(f"Delayed vs Prediction: {cos_sim_dp:.3f}")
```

### 5. Token Overlap Summary

```python
def get_diff_summary(set1, set2):
    return {
        "Common Tokens": len(set1 & set2),
        "Only in Text1": len(set1 - set2),
        "Only in Text2": len(set2 - set1)
    }

summary_id = get_diff_summary(set(tokens_immediate), set(tokens_delayed))
summary_ip = get_diff_summary(set(tokens_immediate), set(tokens_prediction))
summary_dp = get_diff_summary(set(tokens_delayed), set(tokens_prediction))

pd.DataFrame([summary_id, summary_ip, summary_dp],
             index=["Immediate vs Delayed", "Immediate vs Prediction", "Delayed vs Prediction"])
```

### 6. Cosine Similarity Bar Chart

```python
labels = ["Immediate vs Delayed", "Immediate vs Prediction", "Delayed vs Prediction"]
scores = [cos_sim_id, cos_sim_ip, cos_sim_dp]

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, scores, color=['skyblue', 'lightgreen', 'salmon'])
plt.ylim(0, 1)
plt.title("Cosine Similarity Between Transcripts")
plt.ylabel("Similarity Score")
plt.grid(axis='y', linestyle='--', alpha=0.7)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 0.02, f"{height:.2f}", ha='center')
plt.show()
```

###  7. Venn Diagrams

```python
def plot_venn(set1, set2, label1, label2):
    plt.figure(figsize=(6,4))
    venn2([set1, set2], set_labels=(label1, label2))
    plt.title(f"Token Overlap: {label1} vs {label2}")
    plt.show()

plot_venn(set(tokens_immediate), set(tokens_delayed), "Immediate", "Delayed")
plot_venn(set(tokens_immediate), set(tokens_prediction), "Immediate", "Prediction")
plot_venn(set(tokens_delayed), set(tokens_prediction), "Delayed", "Prediction")
```

### 8. Export to CSV

```python
# Save cosine similarity
cos_df = pd.DataFrame({
    "Comparison": ["Immediate vs Delayed", "Immediate vs Prediction", "Delayed vs Prediction"],
    "Cosine Similarity": [cos_sim_id, cos_sim_ip, cos_sim_dp]
})
cos_df.to_csv("cosine_similarity_results.csv", index=False)

# Save token overlap summary
summary_df = pd.DataFrame([summary_id, summary_ip, summary_dp], 
                          index=["Immediate vs Delayed", "Immediate vs Prediction", "Delayed vs Prediction"])
summary_df.to_csv("token_overlap_summary.csv")

print("Saved: cosine_similarity_results.csv and token_overlap_summary.csv")
```
